```python
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
%matplotlib inline

dataset_names = ['Pitt', 'ADReSS', 'Lu']
accuracies = [70.27, 87.50, 90.91]

colors = ['#27F5EE', '#B727F5', '#F5A623']
color_map = dict(zip(['Pitt', 'ADReSS', 'Lu'], colors))
bar_colors = [color_map.get(name, '#888888') for name in dataset_names]

fig, ax = plt.subplots(figsize=(max(8, len(dataset_names)*2), 5))

ax.bar(dataset_names, accuracies, color=bar_colors, alpha=0.7)
ax.set_ylabel('Accuracy (%)', fontsize=12, fontweight='bold')
ax.set_title(f'XLSR Train and Test in Same Datasets', fontsize=13, fontweight='bold')
ax.set_ylim(0, 100)
ax.grid(True, alpha=0.3, axis='y')

for i, v in enumerate(accuracies):
    ax.text(i, v + 2, f'{v:.2f}%', ha='center',fontweight='bold', fontsize=12)

plt.tight_layout()
plt.show()
```
